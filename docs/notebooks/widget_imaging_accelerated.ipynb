# Visualization of Volumetric Imaging Data


```python
from vitessce import (
    VitessceConfig,
    ViewType as vt,
    CoordinationType as ct,
    ImageOmeZarrWrapper,
    CoordinationLevel as CL,
)
from os.path import join
```

## 1. Configure Vitessce
Set up the images from the three different assays, with the `use_physical_size_scaling` set to `True` so that the IMS image scales to the other images based on their physical sizes.

```python
vc = VitessceConfig(schema_version="1.0.18", name='Volume')
dataset = vc.add_dataset(name='Volume').add_object(
    ImageOmeZarrWrapper(img_url='https://data-2.vitessce.io/data/sorger/melanoma_zarr_32/')
)

spatial = vc.add_view("spatialBeta", dataset=dataset).set_props(three=True, accelerated=True)
lc = vc.add_view("layerControllerBeta", dataset=dataset)


vc.layout(spatial | lc);
```

## 2. Create the Vitessce widget

```python
vw = vc.widget()
vw
```
