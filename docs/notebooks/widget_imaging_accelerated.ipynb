# Vitessce Widget Tutorial

# Visualization of Volumetric Imaging Data


```python
from vitessce import (
    VitessceConfig,
    ViewType as vt,
    CoordinationType as ct,
    ImageOmeZarrWrapper,
    CoordinationLevel as CL,
    get_initial_coordination_scope_prefix,
)
```

## 1. Configure Vitessce

Note: OME-Zarr images with cube-shaped chunks work most efficiently
(e.g., prefer a 32^3 chunk shape, rather than chunks like 512x512x1).


```python
vc = VitessceConfig(schema_version="1.0.18", name='Visualization of volumetric data using the spatial-accelerated view')

dataset = vc.add_dataset(name='Volume').add_object(
    ImageOmeZarrWrapper(img_url='https://data-2.vitessce.io/data/sorger/melanoma_zarr_32/')
)

# We currently need to opt-in to the ThreeJS-based accelerated rendering features explicitly via .set_props
spatial_view = vc.add_view("spatialBeta", dataset=dataset).set_props(three=True, accelerated=True)
layer_controller_view = vc.add_view("layerControllerBeta", dataset=dataset)


vc.link_views_by_dict([spatial_view, layer_controller_view], {
    'spatialRenderingMode': '3D',
}, scope_prefix=get_initial_coordination_scope_prefix("A", "image"))


vc.layout(spatial_view | layer_controller_view);
```

## 2. Create the Vitessce widget

```python
vw = vc.widget()
vw
```
